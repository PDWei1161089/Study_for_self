```python
import math
def move(x, y, step, angle):
    nx = x + step * math.cos(angle)
    ny = y - step * math.sin(angle)
    return nx, ny
```

```python
x, y = move(100, 100, 60, math.pi / 6)
print(x)
print(y)
print(x, y)
```

```python
r = move(100, 100, 60, math.pi / 6)
print(r)
```

```python
def fact(n):
    if n==1:
        return 1
    return n * fact(n - 1)
fact(5)
```

```python
def power(x, n):
    s = 1
    while n > 0:
        n = n - 1
        s = s * x
    return s
```

```python
power(5, 2)
```

```python
def power(x, n=2):
    s = 1
    while n > 0:
        n = n - 1
        s = s * x
    return s
```

```python
power(5)
```

```python
def average(*number):
    total = len(number)
    sumn = 0
    for i in number:
        sumn += i
    return sumn / total
```
