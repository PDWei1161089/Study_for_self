## 4.4 浅拷贝与深拷贝
+ 浅拷贝————复制过来的是“引用”，复制对象和被复制对象共享一个共享空间，二者并不是相互独立的
+ 深拷贝————复制过来的是“值”，复制对象和被复制对象占用两个不同空间，两者是相互独立的
+ ndarray的深拷贝方法：.copy()

```python
import numpy as np
myarray1 = np.array(range(0, 10))
myarray2 = myarray1
print(myarray2)
myarray2[1] = 100
print(myarray1)
print(myarray2)
```

```python
import numpy as np
myarray1 = np.array(range(0, 10))
myarray2 = myarray1.copy()
print(myarray2)
myarray2[1] = 200
print(myarray1)
print(myarray2)
```

## 4.7 ndarry的计算
### （1）数组的乘法

```python
myarray5 * 10
```

### (2) 横向拆分

```python
x = np.array([11, 12, 13, 14, 15, 16, 17, 18])
x1, x2, x3 = np.split(x, [3, 5])  # [3,5] 为拆分位置的索引，split是把x在3和5的位置进行切开，切完后元组变成了列表
print(x1, x2, x3)
```

### 纵向拆分, 元组的拆包式赋值

```python
upper, lower = np.vsplit(myarray5.reshape(4, 5), [2])
print("上半部分为\n", upper)
print("下半部分为\n", lower)
```

### 数组的合并

```python
np.concatenate((lower, upper),axis=0)

# axis = 0 表示计算之后列的个数不变，以列为单位进行计算，逐列合并，即上下合并
# axis = 1 表示计算之后行的个数不变，以行为单位进行计算，逐行合并，即左右合并
```

```python
np.vstack([upper, lower])  # 等同于上下合并，即按列合并；列的个数一样
```

```python
np.hstack([upper, lower]) # 等同于左右合并，即按行合并；行的个数一样
```

```python
np.add(myarray5, 1)
```

#### 【注意】：同一个功能，用sum和np.sum()都可以实现，但是：
- 前者为Python内置函数，后者为numpy包的函数，属于ufunc类函数
- ufunc的速度比普通函数快，得益于C语言
- sum为内置函数，并非为ufunc
- np.sum为ufunc函数
#### 【提示】：在Numpy中，数组的函数计算往往为“ufunc类函数”————以列为单位计算的函数，目的是支持向量计算，不用写循环语句

【注意】：np.nan与None的区别：
- None是Python基础语法提供的特殊数据类型，不能参加算术运算

## 4.11 ndarray的广播规则
### 当两个不同构的矩阵运算时，需要用到广播规则
- 规则一：如果列数一样但行数不一样，那么进行以行为单位的广播操作，进行循环补齐

即：低维的向高位广播，列是低维，行是高维，如两位矩阵相加前提是低纬度（列）的维度要相同。

```python
import numpy as np
a1 = np.array(range(1 ,10)).reshape([3, 3])
a1
```

```python
a2 = np.array([10, 10, 10])
a2
```

A1和A2的列数一样，行数不同。

在A1和A2操作之前，进行以行为单位的广播操作，将A1和A2转换为相同结构后进行计算

```python
a1 + a2
```

- 规则二：如果列数不一致（除了列数为1），则解释器将报错

```python
a3 = np.arange(10).reshape(2, 5)
a3
```

```python
a4 = np.arange(16).reshape(4, 4)
a4
```

```python
a3 + a4 # 报错，两者列数不相等
```

# 5.Series

系列，Python中Series, 主要用来处理数据

Pandas包提供了一种key-value型数据结构，其中key为用户定义的显式index,

series和Python列表的区别在于，前者的每个元素都带有两种index，

- 显示index: 由用户指定
- 隐式index: 由系统自动分配

Series的访问既可以通过显式index，也可以通过隐式index。

## 5.1 主要特点

**特点之一**：Series是一种key-value型数据结构，每个元素由如下两部分组成。

- key: 多个显示index
- value: 每个显式index对应的值value

**特点之二**：Series有两种index

- 显式index: 定义一个Series对象时，显式地指定index
- 隐式index: Series对象中每个元素的下标类似Python“列表”的下标

【补充】：

Pandas 中的Series改进dict

dict对集合的改进是有了显式索引

Series 对dict的改进是又加入了下标

## 5.2 定义方法

用pd.Series()函数，参数如下：

- data参数：对应的是Values

        取值：字典、列表等

- index参数：对应的是显式index

        取值：与第一个参数等长的字典、列表等

        注意：若index为字符串时，需要用双引号或单引号括起来，否则会报错。

```python
import pandas as pd
myseries1 = pd.Series([11, 12, 13, 14, 15, 16, 17],index = ["a", "b", "c", "d", "e", "f", "g"])
myseries1
```

```python
import pandas as pd
myseries1 = pd.Series([11, 12, 13, 14, 15, 16, 17],index = [a, b, c, d, e, f])
myseries1
#【注意】报错信息如下，原因分析：index为字符串时，忘记用引号括起来
```

```python
myseries2 = pd.Series([10], index = ["a", "b", "c", "d", "e", "f", "g"])
myseries2 #【注意】：data和index的参数个数要相等
```

```python
myseries3 = pd.Series([1, 2, 3, 4, 5], index=["a", "b", "c"])
myseries3
```

### （6）更新显式index的方法：reindex()

```python
myseries4 = pd.Series([21, 22, 23, 24, 25, 26, 27], index=["a", "b", "c", "d", "e", "f", "g"])
myseries5 = myseries4.reindex(index=["b", "c", "a", "d", "e", "g", "f"])
myseries5
# reindex 重新索引 只是把索引的顺序改变，但是不改变原索引与原数据的对应关系。即通过显式索引修改隐式索引
```

```python
myseries5 = myseries4.reindex(index=["b", "c", "a", "d", "e", "g", "f"])
myseries4
```

【注意】：“更新显式index”是指调整index的次序，而不是定义一个全新的index，即更新后的index必须为已存在，否则自动增加key（显式index）,对应的value为NaN缺失值。

```python
myseries5 = myseries4.reindex(index=["new1", "c", "a", "new2", "e", "g", "new3"])
myseries5
```

【注意】：reindex(不改变Series对象本身的显式index)

## 6.6 算术运算
###### 规则之一：数据框之间的计算规则————先补齐显式index(索引) (新增索引对应值为NaN)， 得到相同结构后，再进行计算。

```python
import pandas as pd
import numpy as np
df4 = pd.DataFrame(np.arange(6).reshape(2, 3))
df4
```

```python
df5 = pd.DataFrame(np.arange(10).reshape(2, 5))
df5
```

```python
df4 + df5  # 先补齐显式索引
```

###### 规则之二：用算术运算符+、-、*等会产生NaN值，如果想将默认值填充的NaN改为指定值，建议不要使用算术运算符，而改用
###### 成员方法，如add()、sub()、mul()、div()。

```python
df6 = df4.add(df5,fill_value = 10)
df6
```

###### 规则之三：数据框与Series的计算规则————按行（第1轴）广播，先把行改为等长，行内不做循环补齐。只是一行一行计算，不会跨行广播。

```python
s1 = pd.Series(np.arange(3))
s1
```

```python
df6 - s1  # 带有显式索引的计算，先补齐显式索引，再进行计算
```

```python
df5 = pd.DataFrame(np.arange(10).reshape(2, 5))
s1 = pd.Series(np.arange(3))
df5 - s1
```

###### 【提示】：等价于成员方法sub()中的axis参数取值为“1”。四则算法对应方法分别为add()、sub()、mul()、div()。

```python
df5 = pd.DataFrame(np.arange(10).reshape(2, 5))
print(df5, "\n")
s1 = pd.Series(np.arange(3))
print(s1)
df5.sub(s1, axis=1)
```

###### 【注意】：axis = 1 的含义可以理解为以下3个条件同时满足。
- (1) 计算前后的行数不变
- (2) 以行为单位进行计算
- (3) 逐行计算
改为按列 (第0轴) 计算的方法：用成员方法/函数，而不用运算符。
###### 【注意】：计算规则：先把列做成等长，再以显式index为依据进行运算

```python
df5 = pd.DataFrame(np.arange(10).reshape(2, 5))
print(df5, "\n")
s1 = pd.Series(np.arange(3))
print(s1)
df5.sub(s1, axis=0)
```

## __6.12 分组统计__
##### groupby的用法如下：
##### (1) 后接一个圆括号和一个方括号
- 圆括号中为分组条件
- 方括号中为计算对象
##### (2) 再接一个方法，如sum()和mean()
##### 即：先按()里的数据分组，再对[]中的内容按后面的方式进行统计

```python
import pandas as pd
df2 = pd.read_csv("bc_data.csv")
df2 = df2[["id", "diagnosis", "area_mean"]]
df2.head()
```

```python
df2.groupby("diagnosis")["area_mean"].mean()
```

```python
# 如果想同时计算多个函数的值，那么需要用方法.aggregate(0,将多个函数名以列表形式枚举在方法aggregate()的参数中
df2.groupby("diagnosis")["area_mean"].aggregate(["mean", "sum", "max", np.median])
```

```python
df2.groupby("diagnosis")["area_mean"].aggregate(["mean", "sum", "max", np.median]).unstack()
# 此处，unstack()函数的功能为“将关系表转换为二级索引”
# stack()和unstack()是数据分析或数据科学项目中常用函数，用于建立和撤销多级索引
```

##### 【提示】：如果将第三个位置上的函数替换成自定义函数，则用方法apply()

```python
def myfunc(x):
    x["area_mean"] /= x["area_mean"].sum()
    return x

df2.groupby("diagnosis").apply(myfunc).head()
```

## __7.3 转换方法__
##### 【提示】：
##### 在人们的日常生活中，表示日期和时间的方式有很多种，如“3th of July.2018”、“2019-1-3”和“2018-07-03 00:00:00”等
- 前两种属于非标准格式，不能作为dt.datetime()的参数
- 最后一种属于标准格式，可以作为dt.datetime()的参数
##### 将非标准格式的时间数据转换为标准格式的时间数据的方法有两种
- 第一种方法：用dateutil包中的parser.parse()
- 第二种方法：用pandas包中的方法to_datetime()

```python
# 例1
dt.datetime("3th of July, 2020")
# 【提示】报错
# 【提示】报错信息为【TypeError: an integer is required (got type str)】
# 原因分析："3th of July, 2020"为【非标准格式】时间数据，不能作为dt.datetime()
```

```python
# 例2
dt.datetime("2019-1-3")
## 【提示】报错信息为【TypeError: an integer is required (got type str)】
# 原因分析：“2019-1-3”为【非标准格式】时间数据，不能作为dt.datetime()的参数
```

### (1) 用dateutil包中的parser()进行日期格式的转换。

```python
from dateutil import parser
date = parser.parse("3th of July, 2018")  # parser可以将自然语言的年月日转换为计算机语言
print(date)
```

```python
date = parser.parse("2019-01-03")
print(date)
```

### (2) 用pandas包中的to_datetime()进行格式转换。 

```python
import pandas as pd
pd.to_datetime("3th of July, 2018")
```

```python
import pandas as pd
pd.to_datetime("2019-1-3")
```

## __8.4 改变图的坐标轴的取值范围__
##### 横坐标和纵坐标的定义：分别采用plt.xlim()和plt.ylim()
##### 横坐标和纵坐标同时定义：plt.axis()

```python
import matplotlib.pyplot as plt
import numpy as np
%matplotlib inline
# 生成试验数据x, 用于后续可视化处理
x = np.linspace(0, 10, 100)
# "np.linspace(0, 10, 100)" 的功能是返回一个含有100个元素且每个元素的取值范围为[0,10]的等距离数列
plt.plot(x, np.sin(x))
plt.xlim(11, -2)  # x轴的取值范围为[11, -2]
plt.ylim(2.2, -1.3)  # y轴的取值范围为[2.2, -1.3]
```

##### 横坐标和纵坐标同时定义：plt.axis(a1, a2, b1, b2)
##### 其中，a1和a2为x轴的取值范围，b1和b2为Y轴的取值范围。

```python
plt.plot(x, np.sin(x))
plt.axis([-1, 21, -1.6, 1.6])
```

```python
plt.plot(x, np.sin(x))
plt.axis([-1, 21, -1.6, 1.6])
plt.axis("equal")
# plt.axis("equal") 的含义为x轴和y轴的刻度单位一样
```

## __8.9 散点图的画法__
##### 【提示】
##### 首先，生成即将用于可视化的试验数据集X和Y。
##### 在本例中使用的生成方法：Python的sklearn.datasets.samples_generator模块提供的函数make_blobs()
##### (1) make_blobs: 生成符合正态分布的随机数据集
- 参数：
#####     n_samples: 样本数量，即行数
#####     n_features:每个样本的特征数量，即列数
#####     centers: 类别数
#####     random_state: 随机数的生成方式
#####     cluster_std: 每个类别的方差
- 返回值，有两个：
#####     X: 测试集，类型为数组，形状为[n_samples, n_features]
#####     Y: 每个成员的标签(label), 也是个数组，形状为[n_samples]的数组
##### (2) plt.scatter函数的参数如下：
#####     X[:,0]和X[:,1]分别为x坐标和y坐标，分别是读取数据框X的第0列和第1列
#####     c为颜色
#####     s为点的大小
#####     cmap为色带，是c的补充

```python
from sklearn.datasets import make_blobs
X, Y = make_blobs(n_samples = 300, centers = 4, random_state = 0, cluster_std = 1.0)
plt.scatter(X[:,0], X[:,1], c = Y, s = 50, cmap = "rainbow")
```

## __8.10 Pandas 可视化__
##### 在数据分析与数据科学项目中，还可以直接用Pandas的画图函数，它继承和优化了Matplotlib，是DataFrame(数据框)类数据的可视化处理更容易
##### Pandas中可以对数据框进行可视化，具体方法：数据框名.plot()。该方法的参数kind决定图的类型。

```python
import pandas as pd
women = pd.read_csv("women.csv", index_col = 0)
women.plot(kind = "bar")
plt.show()
```

```python
women.plot(kind = "barh")  # barh代表的是横向柱状图
plt.show()
```

```python
women.plot(kind="bar", x = "height", y = "weight", color = "g")
plt.show()
```

```python
women.plot(kind = "kde")  # kde代表的是核密度估计曲线
plt.show()
```

```python
women.plot(kind="bar", x="height", y="weight", color="g")
plt.legend(loc="best")  # 图例位置为“最优”
plt.show()
```

## __8.11 Seaborn可视化__
##### Seaborn可以实现比Matplotlib更美观的可视化效果，但需要与Matplotlib结合使用

```python
import matplotlib.pyplot as plt
plt.style.use("classic")  # 设置背景样式
%matplotlib inline
import numpy as np
import pandas as pd
```

```python
# 用plt.style.available查出可用样式列表如下
plt.style.available
```

```python
# 数据准备
rng = np.random.RandomState(0)
x = np.linspace(0, 10, 500)  # 生成500个0~10之间的数
y = np.cumsum(rng.randn(500, 6), 0)
```

##### 【提示】
##### .cumsum是Matlab中的一个函数，通常用于计算一个数组各行的累加值，语法为：
##### B = cumsum(A, dim) 或 B = cumnum(A)

```python
plt.plot(x, y)
plt.legend("abcdef", ncol = 2, loc = "upper left")
```

```python
# 用Seaborn更美观
import seaborn as sns
sns.set()
plt.plot(x, y)
plt.legend("abcdefg", ncol = 2, loc = "upper left")
```

##### 【提示】：
##### plt.legend()的功能为设置图例参数，分别为图例内容("abcdef")、图例列数(ncol = 2)和图例的显示位置(loc = "upper left")

```python
# 仅使用seaborn画图
import pandas as pd
import seaborn as sns
sns.set(style = "ticks")
df_women = pd.read_csv("women.csv", index_col = 0, header = 0)
sns.lmplot(x = "height", y = "weight", data = df_women)
```
